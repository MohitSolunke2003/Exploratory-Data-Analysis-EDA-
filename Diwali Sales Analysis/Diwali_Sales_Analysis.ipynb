# Import Important Python Libries

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
```

# Load the dataset

```python
df = pd.read_csv("Diwali Sales Data.csv", encoding = 'unicode_escape')
```

```python
df.head()
```

# Perform basics stucture operation on data to understand the dataset

```python
df.shape
```

```python
df.info()
```

```python
df.drop(['Status','unnamed1'], axis = 1, inplace = True)
```

```python
df.info()
```

# Handling the Null Values

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
```

```python
df.isnull().sum()
```

# Making Required Changes on Dataset

```python
#Change the data type
df['Amount'] = df['Amount'].astype('int64')
```

```python
df['Amount'].dtypes
```

```python
df.columns
```

```python
#rename the column name
df.rename(columns = {'Marital_Status' : 'Shaadi'})
```

```python
df.describe()
```

```python
#use describe for specfics columns\
df[['Age', 'Orders', 'Amount']].describe()
```

# Exploratory Data Analysis

# Gender

```python
ax = sns.countplot(x = 'Gender', data = df)

for bar in ax.containers:
  ax.bar_label(bar)
```

```python
df.groupby(['Gender'], as_index = False)['Amount'].sum().sort_values(by = 'Amount', ascending= False)
```

```python
sales_gen = df.groupby(['Gender'], as_index = False)['Amount'].sum().sort_values(by = 'Amount', ascending= False)

sns.barplot(x = 'Gender', y = 'Amount', data = sales_gen)
```

# Age

```python
df.columns
```

```python
ax = sns.countplot(x = 'Age Group',data = df, hue = 'Gender')

for bar in ax.containers:
  ax.bar_label(bar)
```

```python
# Total Amount vs Age Group
sales_gen = df.groupby(['Age Group'], as_index = False)['Amount'].sum().sort_values(by = 'Amount', ascending= False)

sns.barplot(x = 'Age Group', y = 'Amount', data = sales_gen)
```

# State

```python
df.columns
```

```python
# Total Number of orders from Top 10 States
sales_state = df.groupby(['State'], as_index= False)['Orders'].sum().sort_values(by = 'Orders', ascending = False).head(10)

sns.set(rc = {'figure.figsize':(15,5)})
sns.barplot(data = sales_state, x = 'State', y = 'Orders')
```

```python
# Total amount of sales/amount from Top 10 State
# Total Number of orders from Top 10 States
sales_state = df.groupby(['State'], as_index= False)['Amount'].sum().sort_values(by = 'Amount', ascending = False).head(10)

sns.set(rc = {'figure.figsize':(15,5)})
sns.barplot(data = sales_state, x = 'State', y = 'Amount')

```

#Marital Status

```python
ax = sns.countplot(x = "Marital_Status", data = df)

sns.set(rc = {'figure.figsize':(7,5)})
for bar in ax.containers:
  ax.bar_label(bar)
```

```python
sales_Marital = df.groupby(['Marital_Status','Gender'], as_index = False)['Amount'].sum().sort_values(by = 'Amount', ascending= False)

sns.barplot(x = 'Marital_Status', y = 'Amount', data  =  sales_Marital, hue = 'Gender')
```

# Occupation

```python
sns.set(rc = {'figure.figsize': (20,5)})
ax = sns.countplot(data = df, x = 'Occupation')

for bar in ax.containers:
  ax.bar_label(bar)
```

from the above graph we can see that most of the buyers are working in IT, Healthcare, Aviation sector

```python
# Occupation vs Amount
sales_occpation = df.groupby(['Occupation'], as_index= False)['Amount'].sum().sort_values(by = 'Amount', ascending= False)

sns.set(rc = {'figure.figsize': (20,5)})
sns.barplot(x = 'Occupation', y = 'Amount', data = sales_occpation)
```

from the above graph we can see that most buyers are working in IT, Healthcare and Aviation Sector

```python
# Occupation vs Amount
sales_occpation = df.groupby(['Product_Category'], as_index= False)['Amount'].sum().sort_values(by = 'Amount', ascending= False)

sns.set(rc = {'figure.figsize': (20,5)})
sns.barplot(x = 'Product_Category', y = 'Amount', data = sales_occpation)
```

#Product Category

```python
sns.set(rc = {'figure.figsize':(20,5)})
ax= sns.countplot(x = 'Product_Category', data= df)

for bar in ax.containers:
  ax.bar_label(bar)
```

From above graph we can see that most of sales are happens of Food, Electronics and clothings categories

```python
# Occupation vs Amount
sales_occpation = df.groupby(['Product_Category'], as_index= False)['Amount'].sum().sort_values(by = 'Amount', ascending= False).head(10)

sns.set(rc = {'figure.figsize': (20,5)})
sns.barplot(x = 'Product_Category', y = 'Amount', data = sales_occpation)
```

- From above graph we can predict that most of the people are spending amount in Food, Clothing and Electronics & Garbage

- But if we can see in above graph most of sales are in clothing but the most amount are earn by food category


# Product ID

```python
# We have to see the which products make a large sales now # Tp 10 Products

sales_product = df.groupby(['Product_ID'], as_index= False)['Amount'].sum().sort_values(by = 'Amount', ascending = False).head(10)

ax = sns.barplot(x = 'Product_ID', y = 'Amount', data = sales_product)

for bar in ax.containers:
  ax.bar_label(bar)
```

```python
# Top 10 most sales products (# Same as )

fig1, ax1 = plt.subplots(figsize = (12,7))
df.groupby('Product_ID')['Amount'].sum().nlargest(10).sort_values(ascending= False).plot(kind='bar')
```
